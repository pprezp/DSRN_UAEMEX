# Perceptrón y sus funciones de activación

En este ejercicio programarás el procesamiento que realiza el perceptrón y las siguientes funciones de activación:

- Escalón unitario
- Sigmoide
- Tangente hiperbólica
- Relu



```python
import numpy as np

# Definición de la función de Escalón Unitario
#   Las salidas de esta funcion se define por la validacion de la suma ponderada recibida
#   Si h >= 0 entonces la salida es 1
#   Si h < 0 entonces la salida es 0
def escalon_unitario(sumaPonderada):
    return 0 if ( sumaPonderada < 0 ) else 1

# Definición de la función sigmoide
def sigmoide(sumaPonderada):
    return 1 / ( 1 + np.exp( -sumaPonderada ) )

# Definición de la función Tangente Hiperbólica
def tangente_hiperbolica( sumaPonderada ):
    return np.tanh( sumaPonderada )

# Definición de la función Relu
#   La salida es el máximo entre 0 y la suma ponderada
def relu( sumaPonderada ):
    return max( 0, sumaPonderada )
```

# Procesamiento dentro del perceptrón

En este bloque programaremos el procesamiento de un perceptrón. Las 2 tareas importantes son:



## Compuerta OR

```python
# Definición de las entradas de la tabla de verdad
entradas = np.array( [ [ 0, 0 ], [ 1, 0 ], [ 0, 1 ], [ 1, 1 ] ] )

# Definión de los resultados esperados del procesamiento de la información por el perceptrón
resultadoEsperado = np.array( [ 0, 1, 1, 1 ] )

# Variable donde se almacenaran los resultados obtenidos
salidas = []

#
pesos = np.array( [ 1, 1 ])
bias = -1

# Para cada entrada ejecuta el procedimiento del perceptrón almacena el resultado en el arreglo de las salidas
for i in range( resultadoEsperado.shape[0] ):
    h = np.dot( entradas[i], pesos ) + bias
    y = escalon_unitario( h )
    salidas.append( [ h, y ] )

# Imprimiendo la salida
print( 'Input1', 'Input2', 'Combinación Lineal', 'Target', 'Salida' )

for i in range( 4 ):
    print( entradas[i][0], entradas[i][1], salidas[i][0], '    ' + str( salidas[i][1] ), resultadoEsperado[i], sep='         '  )
```

## Compuerta AND

```python
# Definición de las entradas de la tabla de verdad
entradas = np.array( [ [ 0, 0 ], [ 1, 0 ], [ 0, 1 ], [ 1, 1 ] ] )

# Definión de los resultados esperados del procesamiento de la información por el perceptrón
resultadoEsperado = np.array( [ 0, 0, 0, 1 ] )

# Variable donde se almacenaran los resultados obtenidos
salidas = []

#
pesos = np.array( [ 1, 2 ])
bias = -3

# Para cada entrada ejecuta el procedimiento del perceptrón almacena el resultado en el arreglo de las salidas
for i in range( resultadoEsperado.shape[0] ):
    h = np.dot( entradas[i], pesos ) + bias
    y = escalon_unitario( h )
    salidas.append( [ h, y ] )

# Imprimiendo la salida
print( 'Utilizando la funcion de activación ESCALÓN UNITARIO ' )
print( 'Input1', 'Input2', 'Combinación Lineal', 'Target', 'Salida' )

for i in range( 4 ):
    print( entradas[i][0], entradas[i][1], salidas[i][0], '    ' + str( salidas[i][1] ), resultadoEsperado[i], sep='         '  )

# Utilizando otra función de activación, la más cercana a probar es RELU con los valores originales
salidas = []

# Para cada entrada ejecuta el procedimiento del perceptrón almacena el resultado en el arreglo de las salidas
for i in range( resultadoEsperado.shape[0] ):
    h = np.dot( entradas[i], pesos ) + bias
    y = relu( h )
    salidas.append( [ h, y ] )

# Imprimiendo la salida
print( '\nUtilizando la funcion de activación RELU con los valores originales' )
print( 'Input1', 'Input2', 'Combinación Lineal', 'Target', 'Salida' )

for i in range( 4 ):
    print( entradas[i][0], entradas[i][1], salidas[i][0], '    ' + str( salidas[i][1] ), resultadoEsperado[i], sep='         '  )

# Utilizando otra función de activación, la más cercana a probar es RELU con los valroes modificando el BIAS
pesos = np.array( [ -2, -1 ])
salidas = []

# Para cada entrada ejecuta el procedimiento del perceptrón almacena el resultado en el arreglo de las salidas
for i in range( resultadoEsperado.shape[0] ):
    h = np.dot( entradas[i], pesos ) + 3
    y = relu( h )
    salidas.append( [ h, y ] )

# Imprimiendo la salida
print( '\nUtilizando la funcion de activación RELU con los valores modificados' )
print( 'Input1', 'Input2', 'Combinación Lineal', 'Target', 'Salida' )

for i in range( 4 ):
    print( entradas[i][0], entradas[i][1], salidas[i][0], '    ' + str( salidas[i][1] ), resultadoEsperado[i], sep='         '  )
```

Lo que podemos validar, es que la función RELU al evaluar el maximo entre 0 y el valor total de la suma ponderada
No es totalmente viable para realizar este tipo de validaciones, ya que nosotros esperamos valores entre 0 y 1
y al tener valores superiores a 1 y el limite inferior ser 0, el rango de validacion incrementa por lo que se necesitaría 
el uso de otro tipo de validaciones o inclusive un perceptrón mas que haga una validación adicional.
