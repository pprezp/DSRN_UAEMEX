```python
import numpy as np

def escalon(x):
    return 1 if x >= 0 else 0

# Perceptrón OR
def OR(x):
    w = np.array([1,1])
    b = -0.5
    return escalon(np.dot(x, w) + b)

# Perceptrón NAND
def NAND(x):
    w = np.array([-1,-1])
    b = 1.5
    return escalon(np.dot(x, w) + b)

# Perceptrón AND
def AND(x):
    w = np.array([1,1])
    b = -1.5
    return escalon(np.dot(x, w) + b)

# XOR usando los anteriores
def XOR(x):
    salida_or = OR(x)
    salida_nand = NAND(x)
    return AND(np.array([salida_or, salida_nand]))

# Prueba
inputs = np.array([[0,0],[0,1],[1,0],[1,1]])

print("x1 x2 -> XOR")
for x in inputs:
    print(x[0], x[1], "->", XOR(x))
```

## Resolviendo XOR con 2 perceptrones vistos anteriormente utilizando aprendizaje con scikit learn


```python
import numpy as np
from sklearn.linear_model import Perceptron

# Definimos las entradas de las tablas de verdad
inputs = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])

#Definimos las salidas de las compuertas lógicas esperadas
targetOR    = np.array([0, 1, 1, 1])
targetNAND  = np.array([1, 1, 1, 0])
targetXOR   = np.array([0, 1, 1, 0]) 

# Definiendo los perceptrones de las diferentes compuertas lógicas
perceptronOR    =   Perceptron( eta0=0.1, max_iter=100, warm_start=True )
perceptronNAND  =   Perceptron( eta0=0.1, max_iter=100, warm_start=True)
perceptronAND   =   Perceptron( eta0=0.1, max_iter=100, warm_start=True )

# Entrenando las compuertas OR y NAND
# Se utiliza fit en lugar de partial_fit, debido a que las iteraciones con partial_fit se tienen que 
# realizar de manera manual, no se toma del parametro max_iter aunque este se defiina.
perceptronOR.fit(inputs, targetOR)
perceptronNAND.fit(inputs, targetNAND)

# Obteniendo las predicciones de OR y NAND
prediccionOR    = perceptronOR.predict(inputs)
prediccionNAND  = perceptronNAND.predict(inputs)

#Generando las entradas para la compuerta AND
inputsAND = np.column_stack((prediccionOR, prediccionNAND))

# Entrenando las compuertas OR y NAND
# Se utiliza fit en lugar de partial_fit, debido a que las iteraciones con partial_fit se tienen que 
# realizar de manera manual, no se toma del parametro max_iter aunque este se defiina.
perceptronAND.fit(inputsAND, targetXOR)

# Obteniendo las predicciones de OR y NAND
prediccionXOR = perceptronAND.predict(inputsAND)

print("Salida OR:  ", prediccionOR)
print("Salida NAND:", prediccionNAND)
print("Salida XOR: ", prediccionXOR)
```

```python
import numpy as np
from sklearn.linear_model import Perceptron

# Definimos las entradas de las tablas de verdad
inputs = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])

#Definimos las salidas de las compuertas lógicas esperadas
targetOR    = np.array([0, 1, 1, 1])
targetNAND  = np.array([1, 1, 1, 0])
targetXOR   = np.array([0, 1, 1, 0]) 

# Definiendo los perceptrones de las diferentes compuertas lógicas
perceptronOR    =   Perceptron( eta0=0.1, max_iter=10, warm_start=True )
perceptronNAND  =   Perceptron( eta0=0.1, max_iter=10, warm_start=True )
perceptronAND   =   Perceptron( eta0=0.1, max_iter=10, warm_start=True )

epocas = 10

for i in range(epocas):
    # Entrenando las compuertas OR y NAND
    # Se utiliza fit en lugar de partial_fit, debido a que las iteraciones con partial_fit se tienen que 
    # realizar de manera manual, no se toma del parametro max_iter aunque este se defiina.
    perceptronOR.partial_fit(inputs, targetOR, classes=np.array( [ 0, 1] ))
    perceptronNAND.partial_fit(inputs, targetNAND, classes=np.array( [ 0, 1] ))

    # Obteniendo las predicciones de OR y NAND
    prediccionOR    = perceptronOR.predict(inputs)
    prediccionNAND  = perceptronNAND.predict(inputs)

    #Generando las entradas para la compuerta AND
    inputsAND = np.column_stack((prediccionOR, prediccionNAND))

    # Entrenando las compuertas OR y NAND
    # Se utiliza fit en lugar de partial_fit, debido a que las iteraciones con partial_fit se tienen que 
    # realizar de manera manual, no se toma del parametro max_iter aunque este se defiina.
    perceptronAND.partial_fit(inputsAND, targetXOR, classes=np.array( [ 0, 1] ))

    # Obteniendo las predicciones de OR y NAND
    prediccionXOR = perceptronAND.predict(inputsAND)

print("Salida OR:  ", prediccionOR)
print("Salida NAND:", prediccionNAND)
print("Salida XOR: ", prediccionXOR)
```
