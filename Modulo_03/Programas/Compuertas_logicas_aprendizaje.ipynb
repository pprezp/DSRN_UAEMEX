# Actividad de tarea: Compuertas lógicas con aprendizaje autom&aacute;tico

```python
# Definimos la función de activacion para nuestro modelo

def escalon_unitario( sumatoria_ponderada ):
    return 1 if ( sumatoria_ponderada >= 0 ) else 0
```

## Cálculo de la compuerta AND

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

# Definición de las entradas de la tabla de verdad
entradas = np.array( [ [ 0, 0 ], [ 1, 0 ], [ 0, 1 ], [ 1, 1 ] ] )

# Definión de los resultados esperados del procesamiento de la información por el perceptrón
resultadoEsperado = np.array( [ 0, 0, 0, 1 ] )

# Definición de los pesos y el bias iniciales
pesos = np.random.rand(2)
bias = np.random.rand(1)

# Definición del learning_rate y las epocas
learning_rate = 0.5
epocas = 10

# Inicializando los arreglos donde se almacenaran los valores y las epocas correspondientes
errores_x_epoca = []
epocas_tiempo = []

# Iterando sobre x número de épocas
for epoca in range(epocas):
    
    #Se setea en 0 los errores por época
    error_total = 0
    
    # Se imprime el cabecero de la información por epoca
    print(f"\n===== ÉPOCA { epoca + 1 } =====")
    print("x1  x2   target   salida   error")

    # Añadiendo la época correspondiente
    epocas_tiempo.append( epoca + 1)

    # Iterando por entradas
    for entrada in range( len( entradas ) ):

        # Obteniendo el valor de las entradas y el valor esperado 
        x = entradas[ entrada ]
        resultado_esperado = resultadoEsperado[ entrada ]

        # Calculando la suma ponderada de los pesos con sus entradas y pasando el resultado a la función de activación
        sumatoria_ponderada = np.dot( entradas[ entrada ], pesos ) + bias
        resultado = escalon_unitario( sumatoria_ponderada )

        # Calculando el margen de error y el absoluto de los errores por entrada
        error = resultado_esperado - resultado
        error_total += abs(error)

        #Actualizando el valor de los pesos y del bias
        pesos = pesos + learning_rate * error * x
        bias = bias + learning_rate * error 

        # Imprimiendo los valores obtenidos con los esperados
        print(f"{ x[ 0 ]}   { x[ 1 ] }      { resultado_esperado }        { resultado }        { error }")
    print(f'Errores totales en la época: { error_total }')
    
    #añadiendo los errores totales por época para graficar
    errores_x_epoca.append( error_total )

# Imprimiendo la validación de lso datos de pesos y bias finales
# realizando los cálculos.
print("\n===== RESULTADO FINAL =====")

# Iterando por cada par de entradas
for i in range(len(entradas)):
    # Obteniendo la suma ponderada
    h = np.dot(entradas[i], pesos) + bias

    # Obteniendo el valor de la función de activación
    y = escalon_unitario(h)

    # Imprimiendo los resultados
    print(f"Entrada: {entradas[i]} -> Salida: {y}")

# Imprimiendo los valores dinales de pesos y bias
print(f'Pesos finales {pesos}')
print(f'bias final {bias}')

# Procediendo a a graficar los valores de errores totales por épocas
plt.plot( epocas_tiempo, errores_x_epoca )
plt.title('Errores a través de las epocas para la compuerta AND')
plt.ylabel('No. de Errores')
plt.xlabel('Epocas')
plt.xticks( range(1, np.max( epocas_tiempo ) + 1 ) )
plt.yticks(range(0, np.max( errores_x_epoca ) + 1 ) )
plt.autoscale(True)
plt.show()
```
