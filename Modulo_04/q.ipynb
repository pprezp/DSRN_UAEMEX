```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
x = ( -10, 0, 10 )
y = ( 0, 0, 1 )

plt.step(x, y, where='pre')
plt.plot(x, y, 'o')
plt.title('Step Plot')
plt.show()
```

```python
# Definición de la función sigmoide
def sigmoide(sumaPonderada):
    return 1 / ( 1 + np.exp( -sumaPonderada ) )


x = np.linspace(-10, 10, 100)
y = sigmoide(x)
plt.plot(x, y)
plt.title('Función Sigmoide')
plt.xlabel('Suma Ponderada')
plt.ylabel('Salida')
plt.grid()
plt.show()
```

```python
# Definición de la función Tangente Hiperbólica
def tangente_hiperbolica( sumaPonderada ):
    return np.tanh( sumaPonderada )

# Definición de la función Tangente Hiperbólica
def tangente_hiperbolica2( sumaPonderada ):
    return ( np.exp(sumaPonderada) - np.exp(-sumaPonderada) )/ (np.exp(sumaPonderada) + np.exp(-sumaPonderada))

x = np.linspace(-5, 5, 50)
y = tangente_hiperbolica(x)
plt.plot(x, y)
plt.title('Función Tangente Hiperbolica')
plt.xlabel('Suma Ponderada')
plt.ylabel('Salida')
plt.grid()
plt.show()


y2 = tangente_hiperbolica2(x)
plt.plot(x, y2)
plt.title('Función Tangente Hiperbolica 2')
plt.xlabel('Suma Ponderada')
plt.ylabel('Salida')
plt.grid()
plt.show()
```
